```python
a = int(input("Masukkan bilangan pertama: "))
b = int(input("Masukkan bilangan kedua: "))
c = int(input("Masukkan bilangan ketiga: "))

rata_rata = (a + b + c) / 3

print("Rata-rata dari ketiga bilangan adalah:", rata_rata)
```

```python
x = int(input("Masukkan nilai x: "))

if x != 0:
    f = 2 * (x ** 3) + 2 * x + 15 / x
    print("Nilai f(x) =", f)
else:
    print("Error: x tidak boleh nol!")
```

```python
print("=== PROGRAM PENUKARAN NILAI DESIMAL ===")
A = float(input("Masukkan nilai desimal A: "))
B = float(input("Masukkan nilai desimal B: "))
C = float(input("Masukkan nilai desimal C: "))
D = float(input("Masukkan nilai desimal D: "))

print("\n[Kondisi Sebelum Ditukar]")
print(f"A = {A}")
print(f"B = {B}")
print(f"C = {C}")
print(f"D = {D}")

A, B, C, D = A, D, B, C

# c. Output
print("\n[Kondisi Setelah Ditukar (A, D, B, C)]")
print(f"Nilai A baru : {A}")
print(f"Nilai B baru : {B}")
print(f"Nilai C baru : {C}")
print(f"Nilai D baru : {D}")
```

```python
umur = int(input("Umur Anda: "))

if umur <= 5:
    kriteria = "Balita"
elif umur <= 13:
    kriteria = "Anak-anak"
elif umur <= 25:
    kriteria = "Remaja"
elif umur <= 35:
    kriteria = "Dewasa"
elif umur <= 55:
    kriteria = "Orang Tua"
else:
    kriteria = "Lansia"

print("Umur =", umur, "dan kriteria yaitu", kriteria)
```
